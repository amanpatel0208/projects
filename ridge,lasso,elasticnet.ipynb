## algerian forest fires dataset

244 instances that regroup a data of two regions of algeria, namely the bejaia region in north east,and sidi bel-abbes in northwest. 122 each

attribute-

1- date

2- temp

3- Rel. humi

4- wind speed

5- rain

6- fine fuel moisture code

7- duff moisture code

8- drought code

9- initial spread index

10- buildup index

11- fire weather index ( dependent )

12- classes (fire or not fire)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# data cleaning

df.isnull().sum()
```

```python
# records with null value

df[df.isnull().any(axis=1)]
```

```python
# the dataset is divided into two regions, we will make new columns for regions

# 0:'bejaia'
# 1: 'sidi bel abbes'

# we label the regions with 0 and 1

```

```python
df.loc[:122,'Region']=0
df.loc[122:,'Region']=1
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]

# we need to drop this
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
# fix spaces in columsn names
df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
df.info()
```

```python
# change the required columns from object type to int type

df[['month','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
# changing the other columns to float

strings=[feature for feature in df.columns if df[feature].dtypes=='str']
```

```python
for i in strings:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
# lets save the cleaned dataset
df.to_csv('Algerian_forest_fires_dataset_cleaned.csv',index=False)
```

```python
df['Classes'].value_counts()

# we see here that the categories are more beacuase of the spaces
```

```python
# encoding of categories in classes table 
# not fire - 0
# fire - 1
# # there are spaces in the classes thus we get many categories
# so we take the category which contains not fire as 1
# and fire as 0
df1['Classes']=np.where(df1['Classes'].str.contains('not fire'),0,1)
```

```python
df1.head()
```

```python
df1['Classes'].value_counts()
# now we see the problem is solved
```

```python
# density plot for all features
plt.style.use('seaborn-v0_8')
df1.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
sns.set_theme()
df1.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
# percentage for pie chart

percentage=df1['Classes'].value_counts(normalize=True)*100

# 1-not fire
# 0- fire
```

```python
# plotting pie chart

classlabels=['Fire','NOt fire']
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.title('pie chart of classes')
```
